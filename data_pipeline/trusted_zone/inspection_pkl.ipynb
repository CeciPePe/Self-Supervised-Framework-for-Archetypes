```python
import pandas as pd
import numpy as np

def add_floors_per_type(df):
    """Compute br_floors_per_type as the number of floors per building use type."""
    def count_floors_by_type(building_spaces):
        if not isinstance(building_spaces, dict):
            return {}
        floors_count = {}
        for use_type, floors in building_spaces.items():
            if isinstance(floors, dict):
                floors_count[use_type] = len(floors)
            else:
                floors_count[use_type] = 0
        return floors_count

    df['br_floors_per_type'] = df['br__building_spaces_by_floor'].apply(count_floors_by_type)
    return df

def explode_building_types(df):
    exploded_rows = []
    for _, row in df.iterrows():
        floors_per_type = row.get('br_floors_per_type', {})
        use_keys = list(floors_per_type.keys())
        for use in use_keys:
            new_row = row.copy()
            # Handle per-type columns
            for col in [
                'br_floors_per_type',
                'br__building_spaces_by_floor',
                'br__building_spaces_reference',
                'br__building_spaces_detailed_use_type',
                'br__communal_area',
                'br__mean_building_space_effective_year',
                'br__mean_building_space_category',
            ]:
                if col in row and isinstance(row[col], dict) and use in row[col]:
                    new_row[col] = row[col][use]
                else:
                    new_row[col] = np.nan

            # Handle above-ground built area
            if isinstance(row.get('br__above_ground_built_area_by_floor', None), list):
                idx = use_keys.index(use)
                values = row['br__above_ground_built_area_by_floor']
                new_row['br__above_ground_built_area_by_floor'] = values[idx] if idx < len(values) else np.nan
            else:
                new_row['br__above_ground_built_area_by_floor'] = np.nan

            # Handle building footprint
            geom = np.nan
            if isinstance(row.get('br__building_footprint_by_floor', None), list):
                floor_dict = row.get('br__building_spaces_by_floor', {}).get(use, {})
                if isinstance(floor_dict, dict):
                    for floor in floor_dict.keys():
                        for item in row['br__building_footprint_by_floor']:
                            if len(item) > 1 and item[1] == floor:
                                geom = item[2]
                                break
                        if geom is not np.nan:
                            break
            new_row['br__building_footprint_by_floor'] = geom

            new_row['use_type'] = use
            exploded_rows.append(new_row)

    df_exploded = pd.DataFrame(exploded_rows)
    df_exploded['idx'] = df_exploded.reset_index().index + 1
    return df_exploded


df = add_floors_per_type(df)
df_exploded = explode_building_types(df)
df_exploded['br__street_width_main_orientation'].replace([np.inf, -np.inf], 0.0, inplace=True)
df_exploded['br__street_width_main_orientation'] = pd.to_numeric(
    df_exploded['br__street_width_main_orientation'], errors='coerce'
)
df_exploded['br__parcel_main_orientation'] = pd.to_numeric(
    df_exploded['br__parcel_main_orientation'], errors='coerce'
)
df_exploded['br__street_width_main_orientation'] /= df_exploded['br__street_width_main_orientation'].max()
df_exploded['br__parcel_main_orientation'] /= df_exploded['br__parcel_main_orientation'].max()


df_exploded.to_csv("preprocessed_data/08279_br_results_exploded.csv", index=False)
df_exploded.head()
```

```python
import os

def export_geoms_as_images(df, geom_col, idx_col, output_folder, image_size=256):
    """
    Export each geometry in a DataFrame as an individual PNG image of fixed pixel size.

    Parameters:
        df (pd.DataFrame): DataFrame containing geometries.
        geom_col (str): Name of the geometry column.
        idx_col (str): Name of the column to use for file names.
        output_folder (str): Directory where PNG images will be saved.
        image_size (int): Size in pixels (width and height) for each output image.
    """
    os.makedirs(output_folder, exist_ok=True)
    dpi = 100
    inches = image_size / dpi

    for _, row in df.iterrows():
        geom = row[geom_col]
        idx = row[idx_col]
        if geom is not None and not pd.isna(geom):
            gdf = gpd.GeoDataFrame(geometry=[geom])
            fig, ax = plt.subplots(figsize=(inches, inches), dpi=dpi)
            gdf.plot(ax=ax, edgecolor="black", facecolor="lightblue")
            ax.set_xticks([])
            ax.set_yticks([])
            ax.set_axis_off()
            output_path = os.path.join(output_folder, f"{idx}.png")
            plt.savefig(output_path, bbox_inches="tight", pad_inches=0)
            plt.close()
    print(f"Exported {len(df)} images to {output_folder}")

# Usage:
export_geoms_as_images(
    df_exploded,
    geom_col="br__building_footprint_by_floor",
    idx_col="idx",
    output_folder="data/data_br/data_all_images",
    image_size=256
)
```
