```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python

meta_trsa2_path = "../data/data_config_1/meta_trsa2.csv"
df_meta_trsa2 = pd.read_csv(meta_trsa2_path)

plt.figure(figsize=(7, 5))
plt.scatter(df_meta_trsa2["YearBuilt1"], df_meta_trsa2["HEIGHT"], alpha=0.5)
plt.title("meta_trsa2.csv: YearBuilt1 vs HEIGHT")
plt.xlabel("YearBuilt1")
plt.ylabel("HEIGHT")
plt.grid(True)
plt.show()


csv_files = [
    "../data/data_config_1/residential/Residential/meta_trsa_Residential.csv",
    "../data/data_config_1/tertiary/Commercial/meta_trsa_Commercial.csv",
    "../data/data_config_1/tertiary/Cultural/meta_trsa_Cultural.csv",
    "../data/data_config_1/tertiary/Entertainment_venues/meta_trsa_Entertainment_venues.csv",
    "../data/data_config_1/tertiary/Healthcare_and_Charity/meta_trsa_Healthcare_and_Charity.csv",
    "../data/data_config_1/tertiary/Industrial/meta_trsa_Industrial.csv",
    "../data/data_config_1/tertiary/Leisure_and_Hospitality/meta_trsa_Leisure_and_Hospitality.csv",
    "../data/data_config_1/tertiary/Offices/meta_trsa_Offices.csv",
    "../data/data_config_1/tertiary/Singular_building/meta_trsa_Singular_building.csv",
    "../data/data_config_1/tertiary/Sports_facilities/meta_trsa_Sports_facilities.csv",
    "../data/data_config_1/tertiary/Urbanization_land/meta_trsa_Urbanization_land.csv",
    "../data/data_config_1/tertiary/Warehouse_Parking/meta_trsa_Warehouse_Parking.csv",
]

for file in csv_files:
    df = pd.read_csv(file)
    counts = df["second_type"].value_counts().sort_index()
    ax = counts.plot(kind="bar", figsize=(8, 4))
    plt.title(f"Distribution of second_type in {file.split('/')[-2]}")
    plt.xlabel("Second Type (encoded)")
    plt.ylabel("Count")
    plt.xticks(rotation=90)
    plt.tight_layout()
    # Add count labels on top of bars
    for i, v in enumerate(counts):
        ax.text(i, v + 0.5, str(v), ha='center', va='bottom', fontsize=9)
    plt.show()
```
